### Loops

- A `loop` allows me to execute a block of code repeatedly without writing the same instructions multiple times.
- Python provides two main types of loops:
     * `while` loop
     * `for`


#### While loop

- A `while` loop executes a block of code as long as it's condition remains `True`. It's perfect for situations where I don't know many times I need to repeat a task

**Syntax**
`while conditions: # Instructions to repeat`

```python
# Print numbers 1 to 5
count = 1
while count <= 5:
    print(count)
    count += 1 # Updating the iteration variable
```

```python
# Print numbers from 5 to 1
n = 5
while n > 0:
    print(n)
    n -= 1 # Updating the iteration variable
    
```

```python
# Print all even numbers between 2 and 10
even_count = 2
while even_count <= 10:
    print(even_count)
    even_count += 2 # Increment the even_count by 2 to get the next even number in the sequence
```

#### For loop

- A `for` loop iterates over items in a sequence, such as a list, string or a range of numbers.
- Perfect for when I want to iterate a specific number of times

**Syntax**
`for variable in sequence: # Instructions`

```python
# Print the names of students
students = ["Wayne", "Mark", "Sahal", "Jeniffer"]
for pupil in students:
    print(pupil)
```

```python
temps = [-5, 2, 0, 15, -1, 10]

# Write your loop here
for t in temps:
    if t > 0:
        print(str(t) + "C  is above freezing.")
    else:
        print(str(t) + "C is freezing")
```

### Iterate through Strings


```python
numbers = [1, 2, 3, 4, 5]
for digit in numbers:
    print(digit)
```

```python
# Count how many characters the word contains
name = "Wayne" # Think of this as putting the word onto a box
count = 0 # Starts a tally at zero
for letter in name: # Goes through the box one letter at a time
    print(letter) # Shows that letter for each letter
    count += 1 # Adds 1 to the tally
print(count) # Prints the final total: 5
```

- An `infinite loop` occurs when a loop continues executing without reaching a termination condition.

### The `break` Statement

- The `break` statement immediately terminates a loop\.
- Why use this? It allows me to check for a "stop condition" anywhere in the loop, not just the beginning.

```python
numbers = [3, 7, 10, 15, 20]
for number in numbers:
    print(number)

    if number == 10: # == is a comparison operator that checks if the value of the number is equal to 10
        print("Number has been found!")
        break # The break statement is used to exit the loop when the condition is met, preventing further iterations.
```

### The `continue` Statement

- The `continue` statement skips the remaining instruction in the currentt iteration and moves to the next iteration.

```python
numbers = [3, 7, 10, 15, 20]
for number in numbers:
    if number == 10:
        continue # The continue statement is used to skip the current iteration of the loop and move on to the next iteration.

    print(number)
```

### Using `range()`

- A `range()` function generates a sequence of integers.

**Syntax**
`range(Start, Stop, Step)`

Parameters
- `start` - Starting value, Defaults to 0.Also it is included.
- `stop` - Ending boundary, which is excluded.
- `step` - Difference betweeen consecutive values, Defaults to 1

```python
# Print numbers from 1 to 10
for number in range(1, 10, 2):
    print(number)
```

### Loop Patterns

1. Counting
2. Summation
3. Find the maximum
4. Find the minimum

```python
# Counting
numbers = [1, 2, 3, 4, 5]
count = 0

for num in numbers:
    count += 1

print("Count:", count)
```

```python
# Summation
numbers = [1, 2, 3, 4, 5]

total = 0 # The loop starts at total = 0 and goes through each number in numbers and goes incrementing it by each number i.e 0 + 1 = 1, 1 + 2 = 3, 3 + 3 = 6, 6 + 4 = 10, 10 + 5 = 15
for num in numbers:
    total += num

print("Total:", total)
```

```python
# Finding the maximum number in a list
numbers = [1, 2, 3, 4, 5]

max_num = numbers[0] # Guesses that the maximum is the first item: 1
for num in numbers: # It checks each numbrer one by one: 1, then 2, then 3 and so on
    if num > max_num: # Here it asks: "Is this current number bigger than the best number I've found so far"
        max_num = num # If yes, it updates max_num to that bigger number

print("Maximum number", max_num) # After the loop finishes, print("Maximum number", max_num) shows the final biggest number
```

```python
# CHALLENGE: Write a while loop that asks the user to guess a secret number
# Set a secret_number = 7
# Inside the loop, ask for an input
# If they guess correctly, use break to exit and print "You win!".
# If they guess a negative number, use continue to tell them "Positive numbers only!" and ask again without checking if it's the secret number.

secret_number = 7
while True:
    guess_number = int(input("Guess a number: "))
    if guess_number < 0:
        print("Positive numbers only!")
        continue
    if guess_number == secret_number:
        print("You win!")
        break
    print("Try again...")
```
