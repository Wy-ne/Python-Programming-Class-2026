#### Sets

- A `set` is a collection of unique elements.
- Sets are created using the `{}` brackets or set constructor `set()`.

##### Characteristics of Sets

- They are unordered.
- They are mutable
like lists and tuples.
- Has no duplicates.
- Not Indexed - Elements cannot be accessed using positions.

##### Creating an Empty Set

```python
empty_set = set()
print(empty_set)
```

##### Set Methods

- `add()` - adds one element to the set
- `update()` - adds multiple elements to the set
- `remove()` removes an element in a set
- `discard()` - removes an element without raising an error if absent
- `clear()` - removes all elements from a set leaving me with an empty set.

##### Adding and Removing Elements

```python
courses = {"Python", "SQL", "Power BI"}
courses.add("Machine Learning")
courses.update(["Data Analysis", "R", "Java"])
print(courses)
```

##### Set Operations


```python
python_students = {"Asma", "Jane", "Mark", "Wayne"}
sql_students = {"Jeniffer", "Nasra", "Sahal", "Ceci", "Wayne"}

# Union - Returns all the unique elements
print(python_students.union(sql_students))

# Intersection - Returns elements found in both sets
print(python_students.intersection(sql_students))

# Difference - Returns the elements in the first set but not in the second set
print(python_students.difference(sql_students))
print(sql_students.difference(python_students))

# Symmetric Difference - Returns elements found in either set but not in both
print(python_students.symmetric_difference(sql_students))
```

```python
# EXERCISE 7: Create a tuple to store mathematical operators(+,-,*,/) and write a program
# that allows users to choose an operator and perform basic calculations with two numbers
first_number = int(input("Enter your first number: "))
second_number = int(input("Enter your second number: "))
operator = input("Enter your operator: (+, -, *, /)")

if operator == '+':
    result = first_number + second_number
    print(f"Result of {first_number} + {second_number} is {result}")
elif operator == '-':
    if second_number <= first_number:
        result = first_number - second_number
        print(f"Result of {first_number} - {second_number} is {result})")
    else:
        print("First number should be greater tha the second number!")
        result = None
elif operator == '*':
    result = first_number * second_number
    print(f"Result of {first_number} * {second_number} is {result}")
elif operator == '/':
    if second_number != 0:
        result = first_number / second_number
        print(f"Result of {first_number} / {second_number} is {result}")
    else:
        print("Error: Division by zero is not allowed!")
        result = None
else:
    print("Invalid Operator")
```
