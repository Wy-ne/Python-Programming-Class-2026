## Functions

- A `function` is a reusable block of code designed to perform a specific task.

**Syntax**
- Python uses the `def` keyword to define a function

`def function_name():`
       `# Instructions to execute`

### Functions Without Parameters

- A function without parameters doesn't receive values from the code calling it.

```python
def greet():
    print("Welcome to Python Class!")

greet()
```

```python
marks = [70, 80, 90]
average = sum(marks) / len(marks)
print("Average:", average)
```

```python
def calculate_average():
    marks = [70, 80, 90]
    average = sum(marks) / len(marks)
    return average

calculate_average()
```

```python
def calculate_area_of_circle():
    radius = 2
    pi = 3.14
    area = pi * radius **2
    print("Area of circle:", area)

calculate_area_of_circle()
```

### Functions With Parameters

- Sometimes a function needs information to perform a task.
- We can provide that information through `parameters`.

```python
def greet_student(name): # 'name' is the parameter
    print("Welcome", name)

greet_student("Wayne") # 'Wayne' is an argument
```

- Here `name` is a parameter. It allows the function to work different values.

```python
# Calculate the area of a rectangle
def area_of_rectangle(length, width):
    area = length * width
    print("Area", area)

area_of_rectangle(5, 4)
area_of_rectangle(10, 5)
```

### Parameters vs Arguments

- `Parameters` are variable names specified when defining the function.
- `Arguments` are actual values supplied when calling the function.

```python
def area_of_square(side_length):
    area = side_length ** 2
    print("Area of the square:", area)

area_of_square(7)
```

- `side_length` - parameter
- `7` - the argument

### Types of Arguments

#### Positional Arguments

- Values assigned according to their position.

```python
def student_details(name, age):
    print("Student Name: ", name)
    print("Student Age: ", age)

student_details("Wayne", 21)
```

#### Keyword Arguments

- Values are assigned using their parameter names.

```python
def student_details(name, age):
    print("Student Name: ", name)
    print("Student Age: ", age)

student_details(name = "Wayne", age = 21)
```

#### Default Arguments

- A parameter can have a default value that is used when no argument is supplied.

```python
def greet_zindua_student(name, course="Data Science"):
    print(f"{name} is studying {course}")

greet_zindua_student("Wayne", "Data Analysis")
```

### Recursive Functions

- A `recursive function` is a function that calls itself to solve a problem.
- Every recursive function needs two important components:
1. `Base case`: The condition that stops the recursion.
2. `Recursive case`: The part where the function keeps calling itself

#### Example: Calculating Factorial

- The `factorial` is the product of all positive integers up to a given number.

- 5! = 5*4*3*2*1 = 120

5 * factorial(4)
4 * factorial(3)
3 * factorial(2)
2 * factorial(1)
1

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)
print(factorial(5))
```
