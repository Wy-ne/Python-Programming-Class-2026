### Comparison Operators
Comparision operators compare two values and return a Boolean result ('True' or 'False')
- '>' - Greater than
- '<' - Less than
- '>=' - Greater than or equal to
- '<=' - Less than or equal to
- '==' - Equal to
- '!= '- Not equal to


```python
average = 73
print(average >= 70)
```

### Logical Operators

Logical Operators get to combine multiple conditions
- 'and' - Both conditions MUST be True
- 'or' - At least one condition MUST be True
- 'not' - Reverses the Boolean result



```python
assignment_score = 75
final_exam_score = 32

assignment_passed = assignment_score >= 50
final_exam_passed = final_exam_score >= 50
overall_pass = assignment_passed and not final_exam_passed
print("Overall pass:", overall_pass)
```

### Assignment Operators

'Assignment Operators' update the value of an existing variable.
- '+=' - x = x + value
- '-=' - x = x - value
- '*=' - x = x * value
- '/=' - x = x / value
- '//=' - x = x // value
- '%=' - x = x % value
- '**=' - x = x ** value


```python
initial_total = 100
total = initial_total + 50
print(total)
```

```python
x = 10
x += 5
print(x)
```

```python
balance = 1000
# Double the balance by 2
balance **= 2
print(balance)
```

### If Statement

- An 'if' statement executes a block of code ONLY when the condition is True. 

**Syntax**
`if condition:`

#### Instructions to execute when the condition is True

```python
age = int(input("Enter your age: "))

if age >= 18:
    print("You're eligible to vote.")
```

**Key Concepts**
- A `condition` - is an expression that evaluates to 'True' or 'False'.
- `Indentation` - identifies the instructions belonging to the condition
- If the condition is `False`, python skips the instructions inside the block.

### The `If-Else` statement

Sometimes I might need a program to choose between two possible outcomes.An if-else statement provides two possible execution paths:
- The `if` block executes when the condition is True
- The `else` block executes when the condition is False


```python
age = int(input("Enter your age: "))
if age >= 18:
    print("You're eligible to vote.")
else:
    print("You're not eligible to vote.")
```

```python
# Write a program to determine whether the number is even or odd.
number = int(input("Enter your number: "))
if number % 2 == 0:
    print(f"The {number} is even")
else:
    print(f"The {number} is odd")
```

### Chained Conditionals `If-Elif-Else'

- Allows me to choose from more than one possibilities

```python
score = int(input("Enter your score: "))
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
    
```

**Quick note**
- Python evaluates the conditions from top to bottom
- The first condition that evaluates to `True` executes, the remaining branches are skipped.

### Nested Conditionals

- A `nested conditional` is an `if` statement placed inside another conditional statement
- Imagine this: A student wants to access an examination.The student must:
1. Be registered
2. Have paid the required fees
- Both conditions must be satisfied

```python
registered = True
fees_paid = False
if registered:
    if fees_paid:
        print("You're registered and fees are paid.")
    else:
        print("You're registered but fees are not paid.")
else:
    print("You're not registered")
```

### Short-Circuit Evaluation and the Guardian Pattern

- Python evaluates logical expressions from `left` to `right`.
- With `and`, if the first condition is `False`, python does not evaluate the remaining conditions. 
- This is called `short-circuit evaluation`. It can help prevent errors when an operation depends on another condition being satisfied first.

### Prevent Division by Zero


```python
x = 10
y = 0

if y != 0 and x / y > 0:
    print("Passed")
else:
    print("Failed")
```
